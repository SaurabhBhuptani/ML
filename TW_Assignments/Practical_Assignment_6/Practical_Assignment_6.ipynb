```python
import csv
import random
import uuid
import datetime
import math

random.seed(42)

# Configuration
NUM_CUSTOMERS = 50000
NUM_TRANSACTIONS = 200000
START_DATE = datetime.datetime(2024,1,1,0,0,0)
END_DATE = datetime.datetime(2024,12,31,23,59,59)

# Customer segment definitions
CUSTOMER_SEGMENTS = ["New", "Regular", "Loyalty"]
CUSTOMER_SEGMENT_PROBS = [0.2, 0.5, 0.3]

PAYMENT_METHODS = ["Card", "Cash", "Mobile Pay"]
PAYMENT_METHOD_PROBS = [0.6, 0.3, 0.1]

def random_timestamp(start, end):
    delta = end - start
    sec = random.randint(0, int(delta.total_seconds()))
    return start + datetime.timedelta(seconds=sec)

def sample_log_normal(mean, sigma):
    # returns a positive float
    return math.exp(random.gauss(mean, sigma))

# Generate customer profiles
customers = []
for i in range(NUM_CUSTOMERS):
    customer_id = str(uuid.uuid4())
    customer_segment = random.choices(CUSTOMER_SEGMENTS, CUSTOMER_SEGMENT_PROBS)[0]
    customers.append((customer_id, customer_segment))

# Write output file
with open("supermarket_transactions.csv", mode="w", newline="") as f:
    writer = csv.writer(f)
    # Header
    writer.writerow([
        "customer_id",
        "transaction_id",
        "transaction_datetime",
        "total_items",
        "total_amount",
        "payment_method",
        "customer_segment"
    ])

    for t in range(NUM_TRANSACTIONS):
        # pick a customer
        customer_id, customer_segment = random.choice(customers)
        transaction_id = str(uuid.uuid4())
        transaction_datetime = random_timestamp(START_DATE, END_DATE)

        # Total items and amount based on customer segment
        if customer_segment == "New":
            total_items = random.randint(1, 10)
            avg_item_price = random.uniform(2.0, 8.0)
        elif customer_segment == "Regular":
            total_items = random.randint(5, 30)
            avg_item_price = random.uniform(3.0, 12.0)
        else:  # Loyalty
            total_items = random.randint(10, 50)
            avg_item_price = random.uniform(4.0, 15.0)

        # Add some noise to the total amount calculation
        total_amount = round(total_items * avg_item_price * random.uniform(0.8, 1.2), 2)
        if total_amount < 0: total_amount = 0.0 # Ensure non-negative amount

        # Payment method
        payment_method = random.choices(PAYMENT_METHODS, PAYMENT_METHOD_PROBS)[0]

        # Write row
        writer.writerow([
            customer_id,
            transaction_id,
            transaction_datetime.isoformat(),
            total_items,
            total_amount,
            payment_method,
            customer_segment
        ])

print("Generated", NUM_TRANSACTIONS, "transactions into supermarket_transactions.csv")
```
