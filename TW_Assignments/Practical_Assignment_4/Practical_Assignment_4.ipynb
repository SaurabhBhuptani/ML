# 1. Explain the impact of selecting the K value over the different natured datasets.

### a. Noisy Datasets
When **K is small**, the model becomes highly sensitive to noise and outliers. With a **larger value of K**, the influence of noise and outliers is reduced, making the model more robust.
### b. Imbalanced Datasets
In an imbalanced dataset, a **small value of K** may struggle to correctly classify minority class samples because most nearest neighbors belong to the majority class. When **K is large**, the bias toward the majority class becomes even stronger, as most neighbors are likely to come from that class.
### c. Datasets with Clear Separation
For datasets with clearly separated classes, a **small value of K** works well because nearest neighbors are easy to identify. A **large value of K** may overly smooth the decision boundary, which can blur clear class distinctions and lead to misclassification.
### d. Datasets with Overlapping Classes
In datasets with overlapping classes, a **small K** can result in incorrect classification due to local ambiguity. A **larger K** provides more stable predictions by considering more neighbors, leading to improved classification in overlapping regions.
### e. High-Dimensional Datasets
In high-dimensional datasets, distance measures become less meaningful because distances between points tend to be similar. This increases noise and classification errors. Dimensionality reduction is often required, but if not done carefully, it can still lead to overfitting.

# 2. Explain the process of applying KNN in regression problems

### Step 1: Choose the value of K
Select an appropriate value of **K**, which represents the number of nearest neighbors used to make a prediction.
### Step 2: Calculate the distance
Compute the distance between the target point and all points in the dataset. **Euclidean distance** is the most commonly used distance measure in KNN.
### Step 3: Find the nearest neighbors
Sort the calculated distances in ascending order and select the **K closest data points**.
### Step 4: Make the prediction
* **For classification:** Assign the target point to the class that occurs most frequently among the K nearest neighbors.
* **For regression:** Compute the average of the target values of the K nearest neighbors, which becomes the predicted value.

# 3. Can KNN be applied on the unlabelled (unsupervised learning) dataset? If yes, then how. If no, then why?

No, KNN cannot be applied directly to an unlabeled dataset for classification because it assigns a label to a new data point based on existing labeled data. Without a labeled dataset, it is not possible to determine the class of a data point. However, KNN can still be used in an unsupervised manner for tasks such as nearest-neighbor search, feature generation, anomaly and outlier detection, and data preprocessing, but not for classification itself.
